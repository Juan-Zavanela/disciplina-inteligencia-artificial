24. Validação de CPF
(Implemente uma função para validar um CPF brasileiro.)

```python
import re

def validar_cpf(cpf: str) -> bool:
    cpf = re.sub(r'\D', '', cpf)

    if len(cpf) != 11:
        return False

    if cpf == cpf[0] * 11:
        return False

    soma = 0
    for i in range(9):
        soma += int(cpf[i]) * (10 - i)
    resto = soma % 11
    digito_1 = 0 if resto < 2 else 11 - resto

    soma = 0
    for i in range(10):
        soma += int(cpf[i]) * (11 - i)
    resto = soma % 11
    digito_2 = 0 if resto < 2 else 11 - resto

    return cpf[-2:] == f"{digito_1}{digito_2}"

# Teste:
print(validar_cpf("111.111.111-11"))
print(validar_cpf("44444444444"))
print(validar_cpf("123.8456.789-10"))
```

25. Classificação de Números
(Classifique uma lista de números em três categorias: positivos,
negativos e zeros.)

```python
def classificar_numeros(lista):
    resultado = {
        "positivos": [],
        "negativos": [],
        "zeros": []
    }

    for numero in lista:
        if numero > 0:
            resultado["positivos"].append(numero)
        elif numero < 0:
            resultado["negativos"].append(numero)
        else:
            resultado["zeros"].append(numero)

    return resultado

numeros = [10, -5, 0, 3, -1, 0, 7, -22]
classificacao = classificar_numeros(numeros)

print("Positivos:", classificacao["positivos"])
print("Negativos:", classificacao["negativos"])
print("Zeros:", classificacao["zeros"])
```

```python
def eh_anagrama(palavra1, palavra2):
    p1 = palavra1.replace(" ", "").lower()
    p2 = palavra2.replace(" ", "").lower()

    if len(p1) != len(p2):
        return False

    return sorted(p1) == sorted(p2)

print(eh_anagrama("Amor", "Roma"))
print(eh_anagrama("Aro", "Ouvido"))
print(eh_anagrama("Despa cho", "Debo chado"))
print(eh_anagrama("Lista", "Tilas"))
```

30. Jogo da Forca
(Desenvolva um jogo simples da forca onde o usuário deve adivinhar
uma palavra.)

```python
import random

def jogar_forca():

    lista_palavras = ["python", "programa", "computador", "algoritmo", "desenvolvedor"]

    palavra_secreta = random.choice(lista_palavras).lower()

    letras_descobertas = ["_" for _ in palavra_secreta]

    tentativas = 6
    letras_tentadas = []
    print("--- Bem-vindo ao Jogo da Forca! ---")

    while tentativas > 0 and "_" in letras_descobertas:
        print(f"\nPalavra: {' '.join(letras_descobertas)}")
        print(f"Tentativas restantes: {tentativas}")
        print(f"Letras já tentadas: {', '.join(letras_tentadas)}")

        palpite = input("Digite uma letra: ").lower().strip()

        if len(palpite) != 1 or not palpite.isalpha():
            print("Por favor, digite apenas uma única letra válida.")
            continue

        if palpite in letras_tentadas:
            print("Você já tentou essa letra. Tente outra!")
            continue

        letras_tentadas.append(palpite)

        if palpite in palavra_secreta:
            print(f"Boa! A letra '{palpite}' está na palavra.")
            for indice, letra in enumerate(palavra_secreta):
                if letra == palpite:
                    letras_descobertas[indice] = palpite
        else:
            print(f"Que pena! A letra '{palpite}' não está na palavra.")
            tentativas -= 1

    if "_" not in letras_descobertas:
        print(f"\nParabéns! Você venceu! A palavra era: {palavra_secreta.upper()}")
    else:
        print(f"\nVocê perdeu! A palavra correta era: {palavra_secreta.upper()}")

jogar_forca()
```

31. Ordenação por Bolha
(Implemente o algoritmo de ordenação por bolha (Bubble Sort).)

```python
def bubble_sort(lista):
    n = len(lista)

    for i in range(n):
        houve_troca = False

        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
                houve_troca = True

        if not houve_troca:
            break

    return lista

lista_desordenada = [64, 34, 25, 12, 22, 11, 90]
lista_ordenada = bubble_sort(lista_desordenada)

print("Lista Ordenada:", lista_ordenada)
```

32. Jogo da Velha
(Crie um jogo da velha para dois jogadores.)

```python
def exibir_tabuleiro(tabuleiro):
    print("\n")
    print(f" {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]} ")
    print("---|---|---")
    print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]} ")
    print("---|---|---")
    print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]} ")
    print("\n")

def verificar_vitoria(tabuleiro, jogador):
    condicoes = [[3, 4, 5], [6, 7, 8], [0, 3, 6], [1, 4, 7], [2, 5, 8], [0, 4, 8], [2, 4, 6]]

    for condicao in condicoes:
        if tabuleiro[condicao[0]] == tabuleiro[condicao[1]] == tabuleiro[condicao[2]] == jogador:
            return True
    return False

def jogo_da_velha():
    tabuleiro = [str(i) for i in range(1, 10)]
    jogador_atual = "X"
    jogadas = 0

    print("--- Bem-vindo ao Jogo da Velha! ---")
    print("Escolha a posição digitando o número correspondente de 1 a 9.")

    while jogadas < 9:
        exibir_tabuleiro(tabuleiro)

        try:
            escolha = int(input(f"Jogador [{jogador_atual}], escolha uma posição (1-9): ")) - 1
        except ValueError:
            print("Entrada inválida! Digite apenas números de 1 a 9.")
            continue

        if escolha < 0 or escolha > 8:
            print("Posição inexistente! Escolha um número de 1 a 9.")
            continue

        if tabuleiro[escolha] in ["X", "O"]:
            print("Essa posição já está ocupada! Tente outra.")
            continue

        tabuleiro[escolha] = jogador_atual
        jogadas += 1

        if verificar_vitoria(tabuleiro, jogador_atual):
            exibir_tabuleiro(tabuleiro)
            print(f"Parabéns! O Jogador [{jogador_atual}] venceu o jogo!")
            return

        jogador_atual = "O" if jogador_atual == "X" else "X"

    exibir_tabuleiro(tabuleiro)
    print("O jogo terminou em Empate (Deu Velha)!")

jogo_da_velha()
```

34. Contador de Vogais
(Conte o número de vogais em uma string fornecida.)

```python
import unicodedata

def contar_vogais(texto: str) -> int:
    texto_normalizado = unicodedata.normalize('NFD', texto)
    texto_limpo = "".join(c for c in texto_normalizado if unicodedata.category(c) != 'Mn')

    vogais = set("aeiou")
    contador = 0

    for caractere in texto_limpo.lower():
        if caractere in vogais:
            contador += 1

    return contador

frase = "Olá, meu nome é Inteligência Artificial!"
total = contar_vogais(frase)

print(f"O texto fornecido contém {total} vogais.")
```

35. Números Fibonacci Até N
(Gere a sequência de Fibonacci até um número n fornecido pelo
usuário.)

```python
def fibonacci_ate_n(n):
    sequencia = [0, 1]

    if n < 0:
        return "Por favor, digite um número maior ou igual a zero."
    if n == 0:
        return [0]
    if n == 1:
        return [0, 1]

    while True:
        proximo_numero = sequencia[-1] + sequencia[-2]

        if proximo_numero <= n:
            sequencia.append(proximo_numero)
        else:
            break

    return sequencia

try:
    limite = int(input("Digite o valor limite (N) para a sequência: "))
    resultado = fibonacci_ate_n(limite)
    print(f"Sequência de Fibonacci até {limite}: {resultado}")
except ValueError:
    print("Erro: Digite apenas números inteiros válidos.")
```

36. Gerador de Senhas
(Crie um gerador de senhas aleatórias com letras e números.)

```python
import secrets
import string

def gerar_senha(comprimento=12):
    caracteres = string.ascii_letters + string.digits

    if comprimento < 4:
        comprimento = 4

    senha = "".join(secrets.choice(caracteres) for _ in range(comprimento))

    return senha

try:
    tamanho = int(input("Digite o comprimento desejado para a senha (ex: 12): "))
    senha_gerada = gerar_senha(tamanho)
    print(f"\nSua nova senha segura: {senha_gerada}")
except ValueError:
    print("Erro: Digite apenas números inteiros válidos para o comprimento.")
```

```python
def exibir_menu():
    print("\n=== GERENCIADOR DE LISTA DE COMPRAS ===")
    print("1. Adicionar item")
    print("2. Remover item")
    print("3. Listar itens")
    print("4. Sair")
    return input("Escolha uma opção (1-4): ").strip()

def gerenciar_compras():
    lista_compras = []

    while True:
        opcao = exibir_menu()

        if opcao == "1":
            item = input("Digite o nome do item para adicionar: ").strip()
            if item:
                lista_compras.append(item)
                print(f" '{item}' foi adicionado com sucesso!")
            else:
                print(" O nome do item não pode ser vazio.")

        elif opcao == "2":
            if not lista_compras:
                print(" A lista está vazia. Não há nada para remover.")
                continue

            print("\nItens atuais:")
            for indice, item in enumerate(lista_compras, 1):
                print(f"{indice}. {item}")

            try:
                escolha = int(input("Digite o número do item que deseja remover: "))
                if 1 <= escolha <= len(lista_compras):
                    removido = lista_compras.pop(escolha - 1)
                    print(f" '{removido}' foi removido da lista.")
                else:
                    print(" Número inválido! Escolha uma posição existente.")
            except ValueError:
                print(" Entrada inválida! Digite apenas o número do item.")

        elif opcao == "3":
            if not lista_compras:
                print(" Sua lista de compras está vazia.")
            else:
                print("\n = LISTA DE COMPRAS =")
                for indice, item in enumerate(lista_compras, 1):
                    print(f"- {item}")

        elif opcao == "4":
            print(" Saindo do programa. Boas compras!")
            break

        else:
            print("Opção inválida! Tente novamente.")

# Executa o programa
gerenciar_compras()
```

42. Números Primos até N
(Gere todos os números primos até um número n fornecido.)

```python
def gerar_primos_ate_n(n):
    if n < 2:
        return []

    primos = [True] * (n + 1)
    primos[0] = primos[1] = False

    for i in range(2, int(n**0.5) + 1):
        if primos[i]:
            for j in range(i * i, n + 1, i):
                primos[j] = False

    return [numero for numero, eh_primo in enumerate(primos) if eh_primo]

try:
    limite = int(input("Digite o valor limite (N): "))
    resultado = gerar_primos_ate_n(limite)
    print(f"Números primos até {limite}: {resultado}")
except ValueError:
    print("Erro: Digite apenas números inteiros válidos.")
```

43. Conversor de Unidades
(Converta unidades de comprimento (por exemplo, metros para
centímetros).)

```python
def converter_comprimento(valor, da_unidade, para_unidade):
    fatores_para_metro = {
        'm': 1.0,
        'cm': 0.01,
        'mm': 0.001,
        'km': 1000.0
    }

    valor_em_metros = valor * fatores_para_metro[da_unidade]

    valor_final = valor_em_metros / fatores_para_metro[para_unidade]

    return valor_final

def menu_conversor():
    print("--- Conversor de Unidades de Comprimento ---")
    print("Unidades aceitas: m (metros), cm (centímetros), mm (milímetros), km (quilômetros)")

    try:
        da_unidade = input("Converter DE (digite a sigla): ").strip().lower()
        para_unidade = input("Converter PARA (digite a sigla): ").strip().lower()

        unidades_validas = ['m', 'cm', 'mm', 'km']

        if da_unidade not in unidades_validas or para_unidade not in unidades_validas:
            print("Erro: Uma ou ambas as unidades informadas são inválidas.")
            return

        valor = float(input(f"Digite o valor em [{da_unidade}]: "))

        resultado = converter_comprimento(valor, da_unidade, para_unidade)

        print(f"\n Resultado: {valor} {da_unidade} = {resultado:.4f} {para_unidade}")

    except ValueError:
        print("Erro: Por favor, insira um número válido para a conversão.")

menu_conversor()
```

44. Fibonacci em Nível
(Gere a sequência de Fibonacci até um determinado número de
termos.)

```python
def fibonacci_termos(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]

    sequencia = [0, 1]

    while len(sequencia) < n:
        proximo = sequencia[-1] + sequencia[-2]
        sequencia.append(proximo)

    return sequencia

try:
    quantidade = int(input("Digite quantos termos da sequência você deseja gerar: "))
    resultado = fibonacci_termos(quantidade)
    print(f"\n Os primeiros {quantidade} termos da sequência são:")
    print(resultado)
except ValueError:
    print(" Erro: Digite apenas números inteiros válidos.")
```

45. Par ou Ímpar em Lista
(Classifique os números de uma lista como pares ou ímpares.)

```python
def classificar_par_impar(lista):
    resultado = {
        "pares": [],
        "impares": []
    }

    for numero in lista:
        if numero % 2 == 0:
            resultado["pares"].append(numero)
        else:
            resultado["impares"].append(numero)

    return resultado

numeros = [14, 27, 33, 42, 8, 19, 90, 55]
classificacao = classificar_par_impar(numeros)

print("Números Pares:", classificacao["pares"])
print("Números Ímpares:", classificacao["impares"])
```

46. Calcular Média e Desvio Padrão
(Calcule a média e o desvio padrão de uma lista de números.)

```python
def calcular_media_e_desvio(lista):
    n = len(lista)
    if n == 0:
        return 0, 0

    media = sum(lista) / n

    soma_diferencas_quadrado = sum((x - media) ** 2 for x in lista)
    variancia = soma_diferencas_quadrado / n

    desvio_padrao = variancia ** 0.5

    return media, desvio_padrao

dados = [10, 20, 30, 40, 50]
media, desvio = calcular_media_e_desvio(dados)

print(f"Média: {media:.2f}")
print(f"Desvio Padrão: {desvio:.2f}")
```

47. Contar Números em Lista
(Conte quantas vezes cada número aparece em uma lista.)

```python
from collections import Counter

def contar_frequencia(lista):
    return Counter(lista)

numeros = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
resultado = contar_frequencia(numeros)

print("Frequência dos números:")
for numero, quantidade in resultado.items():
    print(f"O número {numero} aparece {quantidade} vez(es).")
```

48. Ordenação por Inserção
(Implemente o algoritmo de ordenação por inserção (Insertion Sort).)

```python
def insertion_sort(lista):
    for i in range(1, len(lista)):
        chave = lista[i]
        j = i - 1


        while j >= 0 and lista[j] > chave:
            lista[j + 1] = lista[j]
            j -= 1


        lista[j + 1] = chave

    return lista

lista_desordenada = [12, 11, 13, 5, 6]
lista_ordenada = insertion_sort(lista_desordenada)

print("Lista Ordenada:", lista_ordenada)
```

49. Soma dos Dígitos
(Calcule a soma dos dígitos de um número inteiro fornecido.)

```python
def soma_digitos_texto(numero):
    numero_str = str(abs(numero))

    return sum(int(digito) for digito in numero_str)

num = 12345
print(f"A soma dos dígitos de {num} é: {soma_digitos_texto(num)}")
```

50. Cálculo de Frequência de Letras
(Calcule a frequência de cada letra em uma string fornecida pelo
usuário.)

```python
from collections import Counter

def calcular_frequencia_letras(texto: str):
    texto = texto.lower()

    letras_filtradas = [caractere for caractere in texto if caractere.isalpha()]

    frequencia = Counter(letras_filtradas)

    return frequencia

texto_usuario = input("Digite um texto para analisar a frequência das letras: ")
resultado = calcular_frequencia_letras(texto_usuario)

print("\n Frequência de cada letra:")
for letra, contagem in sorted(resultado.items()):
    print(f"Letra '{letra}': {contagem} vez(es)")
```
