1. Calcule a integral da fun¸c˜ao f(x) = x2 sin(x) no intervalo de 0 a π.

```python
import numpy as np
from scipy import integrate, optimize, fft, interpolate, linalg

def f1(x):
    return x**2 * np.sin(x)

resultado, erro = integrate.quad(f1, 0, np.pi)

print("Resultado da integral:", resultado)
```

4. Use a fun¸c˜ao de otimiza¸c˜ao de SciPy para maximizar f(x) = −x2+4x+1.

```python
def f4_negativa(x):
    return -(-x**2 + 4*x + 1)

resultado4 = optimize.minimize_scalar(f4_negativa)
x_maximo = resultado4.x
valor_maximo = -resultado4.fun

print("x onde o maximo acontece:", x_maximo)
print("Valor maximo da funcao:", valor_maximo)
```

5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e
cos(4πt).

```python
quantidade_pontos = 500
tempo = np.linspace(0, 1, quantidade_pontos, endpoint=False)
sinal = np.sin(2*np.pi*tempo) + np.cos(4*np.pi*tempo)

transformada = fft.fft(sinal)
frequencias = fft.fftfreq(quantidade_pontos, tempo[1]-tempo[0])

metade = quantidade_pontos // 2
for i in range(metade):
    amplitude = np.abs(transformada[i]) * 2 / quantidade_pontos
    if amplitude > 0.1:
        print(f"frequencia = {frequencias[i]:.2f} Hz | amplitude = {amplitude:.2f}")
```

7. Use a fun¸c˜ao de interpola¸c˜ao c´ubica de SciPy para interpolar os dados
(0, 0), (1,2),(2,3),(3,5),(4,4).

```python
x_pontos = [0, 1, 2, 3, 4]
y_pontos = [0, 2, 3, 5, 4]

spline = interpolate.CubicSpline(x_pontos, y_pontos)

novos_x = np.linspace(0, 4, 9)
novos_y = spline(novos_x)

for x, y in zip(novos_x, novos_y):
    print(f"x = {x:.1f}  ->  y = {y:.2f}")
```
